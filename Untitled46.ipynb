```python
# Change date_of_birth_or_registration from OBJ to DATE format:
import pandas as pd

data = pd.read_csv("Raw Data/santam_raw_customers.csv")

data["date_of_birth_or_registration"] = pd.to_datetime(
    data["date_of_birth_or_registration"],
    errors="coerce"
)

print(data["date_of_birth_or_registration"].dtype)
```

```python
# Check data in 'province' column:
print(data["province"].unique())
```

```python
# Replace all Nan values in 'province' to 'Unspecified'
data["province"] = data["province"].fillna("Unspecified")
```

```python
# Check data in 'customer_type':
print(data["customer_type"].unique())
```

```python
# Capitalise data in 'customer_type' column:
data["customer_type"] = data["customer_type"].str.capitalize()
```

```python
# Check data in 'source_system':
print(data["source_system"].unique())
```

```python
data.head()
```

```python
# Correct wording format in 'source_system' column:
data['source_system'] = data['source_system'].replace(
    'LegacyShortTermB',
    'Legacy Short Term B')

data['source_system'] = data['source_system'].replace(
    'LegacyShortTermA',
    'Legacy Short Term A')

data['source_system'] = data['source_system'].replace(
    'BrokerPortal',
    'Broker Portal')

data['source_system'] = data['source_system'].replace(
    'AcquiredBook',
    'Acquired Book')

```

```python
data.head()
```

```python
# Cleam date foramt in 'date_of_birth_or_registration' column:
data['date_of_birth_or_registration'] = data['date_of_birth_or_registration'].replace(
    '/',
    '-')
```

```python
# Replace NaT with 'Unspecified' in 'date_of_birth_or_registration' column:
data['date_of_birth_or_registration'] = data['date_of_birth_or_registration'].fillna('Unspecified')
```
