## Loss Date Feature

```python
# Split the date into parts
parts = data['loss_date'].astype(str).str.replace('-', '/', regex=False).str.split('/', expand=True)

# Extract components
part1 = pd.to_numeric(parts[0], errors='coerce')
part2 = pd.to_numeric(parts[1], errors='coerce')

# Find obvious invalid values
invalid_dates = data[
    (part1 > 31) |
    (part2 > 31) |
    (part1 == 0) |
    (part2 == 0)
]

print(invalid_dates[['source_row_id', 'loss_date']])
data.head()
```

```python
data['loss_date'].unique()
```

```python
data['loss_date'] = data['loss_date'].replace(
    'not-a-date',
    pd.NA
)
data.tail(5)
```

```python
parsed_loss_date = pd.to_datetime(
    data['loss_date'],
    format='mixed',
    errors='coerce'
)
 
loss_date_quarantine_mask = parsed_loss_date.isna()
claims_quarantine = data[loss_date_quarantine_mask].copy()
claims_working = data[~loss_date_quarantine_mask].copy()
```

## Claim Amount Feature

```python
data['claim_amount'].unique()
```

```python
data['claim_amount'].value_counts(dropna=False)
```

```python
data['claim_amount'] = data['claim_amount'].replace({
    'R --': pd.NA,
    'not recorded': pd.NA
})
```

```python
data['claim_amount'] = (
    data['claim_amount']
    .str.replace('ZAR', '', regex=False)
    .str.replace('R', '', regex=False)
    .str.strip()
)
```

```python
mask = data['claim_amount'].str.contains(
    r'\.\d{3},\d{2}', na=False
)
 
data.loc[mask, 'claim_amount'] = (
    data.loc[mask, 'claim_amount']
    .str.replace('.', '', regex=False)
    .str.replace(',', '.', regex=False)
)
 
data['claim_amount'] = (
    data['claim_amount']
    .str.replace(',', '', regex=False)
)
```

```python
data['claim_amount'] = pd.to_numeric(
    data['claim_amount'],
    errors='coerce'
)
```
