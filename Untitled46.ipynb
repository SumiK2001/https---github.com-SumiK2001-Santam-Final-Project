```python
#Replacing the invalid amount wit NaN
policies['annual_premium'] = policies['annual_premium'].replace(
    ['--', 'not recorded'],
    pd.NA
)

policies['sum_insured'] = policies['sum_insured'].replace(
    ['--', 'not recorded'],
    pd.NA
)
```

```python
remaining_invalid = policies.loc[
    pd.to_numeric(
        policies['annual_premium'],
        errors='coerce'
    ).isna(),
    'annual_premium'
]

remaining_invalid.value_counts()
```

```python
#Removing CURRENCY in the money
policies['annual_premium'] = (
    policies['annual_premium']
    .astype(str)
    .str.replace('ZAR', '', regex=False)
    .str.replace('R', '', regex=False)
    .str.strip()
)

policies['sum_insured'] = (
    policies['sum_insured']
    .astype(str)
    .str.replace('ZAR', '', regex=False)
    .str.replace('R', '', regex=False)
    .str.strip()
)
```

```python
# CSV for all the invalid amounts '--', 'not recorded'
quarantine_amounts = policies[
    policies['annual_premium'].isin(['--', 'not recorded'])
    |
    policies['sum_insured'].isin(['--', 'not recorded'])
]

quarantine_amounts.to_csv(
    "Quarantine/quarantine_invalid_amounts.csv",
    index=False
)
```

```python
# Converting Vehicle  Year to int
policies['vehicle_year'] = policies['vehicle_year'].astype('Int64')
```

```python
# Replacing INFORCE wit In Force in Policy status
policies['policy_status'] = policies['policy_status'].replace({
    'INFORCE': 'IN FORCE'
})
policies.policy_status.unique()
```

```python
# Fixing inconsistancies on Product Line and Policy Status
# Remove spaces and standardize capitalization
cols = ['product_line', 'policy_status']

for col in cols:
    policies[col] = (
        policies[col]
        .str.strip()
        .str.title()
    )

```

```python
# Coveerting Date to date type
policies['policy_start_date'] = pd.to_datetime(
    policies['policy_start_date'],
    format='mixed'
)

policies['policy_end_date'] = pd.to_datetime(
    policies['policy_end_date'],
    format='mixed'
)
policies.info()
```

```python
#Checking invalid dates 

tmp_start = pd.to_datetime(
    policies['policy_start_date'],
    format='mixed',
    errors='coerce'
)

tmp_end = pd.to_datetime(
    policies['policy_end_date'],
    format='mixed',
    errors='coerce'
)

print("Invalid policy_start_date values:")
print(
    policies.loc[
        tmp_start.isna(),
        'policy_start_date'
    ].value_counts(dropna=False)
)

print("\nInvalid policy_end_date values:")
print(
    policies.loc[
        tmp_end.isna(),
        'policy_end_date'
    ].value_counts(dropna=False)
)
 # WHAT i GOT AS RESULTS

# Invalid policy_start_date values:
# policy_start_date
# NaN           77
# 31/31/2024    75
# not-a-date    67
# 2024-19-42    60
# Name: count, dtype: int64

# Invalid policy_end_date values:
# policy_end_date
# 31/31/2024    81
# not-a-date    77
# 2024-19-42    71
# NaN           67
# Name: count, dtype: int64
```

```python
#THE QUARANTINE CSV FOR ALL THE INVALID DATES
# Invalid dates in either column
mask = tmp_start.isna() | tmp_end.isna()

# Create quarantine dataset
quarantine_dates = policies.loc[mask].copy()

# Save quarantine file
quarantine_dates.to_csv(
    "Quarantine/quarantine_all_invalid_dates.csv",
    index=False
)
```

```python
# Rplace invalid dates 

# Replace invalid start dates with NaT
policies.loc[
    tmp_start.isna(),
    'policy_start_date'
] = pd.NaT

# Replace invalid end dates with NaT
policies.loc[
    tmp_end.isna(),
    'policy_end_date'
] = pd.NaT
```
