# Claims

```python
import pandas as pd

data = pd.read_csv("Raw Data/santam_raw_claims.csv")

data.shape
```

```python
import pandas as pd

data = pd.read_csv("Raw Data/santam_raw_claims.csv")

# Number of unique values per column
distinct_values = data.nunique()

print(distinct_values)
```

```python
data.info()
```

```python
# Changing the date format

print(data['loss_date'].head(20).tolist())
```

```python
sample = data['loss_date'].dropna().sample(30, random_state=42)

for x in sample:
    print(x)
```

```python
#Changing date data type
data['settled_date'] = pd.to_datetime(
   data['settled_date'],
    format='mixed',
    dayfirst=True,
    errors='coerce'
)
data.info()
```

```python
invalid_loss_dates = data[data['loss_date'].isna()].copy()

print(invalid_loss_dates.shape)

data.info
```

```python
missing = data.isnull().sum().sort_values(ascending=False)

print(missing)
```

```python
amount_cols = [
    'claim_amount',
    'authorised_amount',
    'paid_amount',
    'outstanding_reserve'
]

for col in amount_cols:
    data[col] = pd.to_numeric(
        data[col],
        errors='coerce'
    )
```

```python
#Changing datatypes on claim_amount, authorised_amount,paid_amount 

data['claim_amount'] = data['claim_amount'].astype(float)
data['authorised_amount'] = data['authorised_amount'].astype(float)
data['paid_amount'] = data['paid_amount'].astype(float)
data.info()
```

```python
data.province = data.province.str.replace(r'\s+', ' ', regex=True)
data.province.unique()
```

```python
data.claim_status = data.claim_status.str.replace(r'\s+', ' ', regex=True)
data.claim_status.unique()
```

```python
data.cause_of_loss = data.cause_of_loss.str.replace(r'\s+', ' ', regex=True)
data.cause_of_loss.unique()
```
