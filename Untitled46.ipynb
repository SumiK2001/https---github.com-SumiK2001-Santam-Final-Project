# Policies

```python
import pandas as pd

policies = pd.read_csv("Raw Data/santam_raw_policies.csv")

```

```python
policies.head()
```

```python
policies.info()
```

```python
#putting a placeholder, Since these fields are not applicable for non-motor policies

cols = [
    'vehicle_year',
    'vehicle_registration',
    'vehicle_vin_placeholder',
    'postal_area',
    'province'
]

policies[cols] = policies[cols].replace('', pd.NA)
```

```python
# Removing , and . on the annual_premium

mask = policies['annual_premium'].str.contains(
    r'\.\d{3},\d{2}', na=False
)
 
policies.loc[mask, 'annual_premium'] = (
    policies.loc[mask, 'annual_premium']
    .str.replace('.', '', regex=False)
    .str.replace(',', '.', regex=False)
)
 
policies['annual_premium'] = (
    policies['annual_premium']
    .str.replace(',', '', regex=False)
)
```

```python
# Removing , and . on the sum_insured

mask = policies['sum_insured'].str.contains(
    r'\.\d{3},\d{2}', na=False
)
 
policies.loc[mask, 'sum_insured'] = (
    policies.loc[mask, 'sum_insured']
    .str.replace('.', '', regex=False)
    .str.replace(',', '.', regex=False)
)
 
policies['sum_insured'] = (
    policies['sum_insured']
    .str.replace(',', '', regex=False)
)
```

```python
# Checking invalid amounts 

invalid_amounts = policies.loc[
    pd.to_numeric(policies['annual_premium'], errors='coerce').isna()
    & policies['annual_premium'].notna(),
    'annual_premium'
]

invalid_amounts.value_counts()
```

```python
#Replacing the invalid amount wit NaN
policies['annual_premium'] = policies['annual_premium'].replace(
    ['--', 'not recorded'],
    pd.NA
)

policies['sum_insured'] = policies['sum_insured'].replace(
    ['--', 'not recorded'],
    pd.NA
)
```

```python
# CONVERTING ANNUAL_PREMIUM AND SUM_INSURED TO FLOAT 
policies['annual_premium'] = pd.to_numeric(
    policies['annual_premium'],
    errors='coerce'
)

policies['sum_insured'] = pd.to_numeric(
    policies['sum_insured'],
    errors='coerce'
)
```

```python
#Removing CURRENCY in the money
policies['annual_premium'] = (
    policies['annual_premium']
    .astype(str)
    .str.replace('ZAR', '', regex=False)
    .str.replace('R', '', regex=False)
    .str.strip()
)

policies['sum_insured'] = (
    policies['sum_insured']
    .astype(str)
    .str.replace('ZAR', '', regex=False)
    .str.replace('R', '', regex=False)
    .str.strip()
)
```

```python
# CSV for all the invalid amounts '--', 'not recorded'
quarantine_amounts = policies[
    policies['annual_premium'].isin(['--', 'not recorded'])
    |
    policies['sum_insured'].isin(['--', 'not recorded'])
]

quarantine_amounts.to_csv(
    "Quarantine/quarantine_invalid_amounts.csv",
    index=False
)
```

```python
#Checking invalid amounts

invalid_amounts = policies.loc[
    pd.to_numeric(policies['annual_premium'], errors='coerce').isna()
    & policies['annual_premium'].notna(),
    'annual_premium'
]

invalid_amounts.value_counts()
```

```python
# Converting Vehicle  Year to int
policies['vehicle_year'] = policies['vehicle_year'].astype('Int64')
```

```python
# Replacing INFORCE wit In Force in Policy status
policies['policy_status'] = policies['policy_status'].replace({
    'INFORCE': 'IN FORCE'
})
policies.policy_status.unique()
```

```python
# Fixing inconsistancies on Product Line and Policy Status
# Remove spaces and standardize capitalization
cols = ['product_line', 'policy_status']

for col in cols:
    policies[col] = (
        policies[col]
        .str.strip()
        .str.title()
    )

```

```python
# Coveerting Date to date type
policies['policy_start_date'] = pd.to_datetime(
    policies['policy_start_date'],
    format='mixed'
)

policies['policy_end_date'] = pd.to_datetime(
    policies['policy_end_date'],
    format='mixed'
)
policies.info()
```

```python
#Checking invalid dates 

tmp_start = pd.to_datetime(
    policies['policy_start_date'],
    format='mixed',
    errors='coerce'
)

tmp_end = pd.to_datetime(
    policies['policy_end_date'],
    format='mixed',
    errors='coerce'
)

print("Invalid policy_start_date values:")
print(
    policies.loc[
        tmp_start.isna(),
        'policy_start_date'
    ].value_counts(dropna=False)
)

print("\nInvalid policy_end_date values:")
print(
    policies.loc[
        tmp_end.isna(),
        'policy_end_date'
    ].value_counts(dropna=False)
)
 # WHAT i GOT AS RESULTS

# Invalid policy_start_date values:
# policy_start_date
# NaN           77
# 31/31/2024    75
# not-a-date    67
# 2024-19-42    60
# Name: count, dtype: int64

# Invalid policy_end_date values:
# policy_end_date
# 31/31/2024    81
# not-a-date    77
# 2024-19-42    71
# NaN           67
# Name: count, dtype: int64
```

```python
#THE QUARANTINE CSV FOR ALL THE INVALID DATES
# Invalid dates in either column
mask = tmp_start.isna() | tmp_end.isna()

# Create quarantine dataset
quarantine_dates = policies.loc[mask].copy()

# Save quarantine file
quarantine_dates.to_csv(
    "Quarantine/quarantine_all_invalid_dates.csv",
    index=False
)
```

# Premium Transactions

```python
import pandas as pd

premium_transactions = pd.read_csv("Raw Data/santam_raw_premium_transactions.csv")

premium_transactions.shape
```

```python
# Columns to check for nulls
check_cols = [
    'transaction_date',
    'cover_start_date',
    'cover_end_date',
    'amount',
    'commission_amount',
    'acquisition_expense_amount',
    'administration_expense_amount'
]

# Rows with at least one null
null_mask = premium_transactions[check_cols].isna().any(axis=1)

# Quarantine dataset
quarantine_nulls = premium_transactions.loc[null_mask].copy()

# List the columns that are null in each row
quarantine_nulls['anomaly_columns'] = (
    quarantine_nulls[check_cols]
    .isna()
    .apply(
        lambda row: ', '.join(row.index[row]),
        axis=1
    )
)
```

```python
# Quarantine all the rows with nulls
date_cols = [
    'transaction_date',
    'cover_start_date',
    'cover_end_date'
]

quarantine_nulls['anomaly_code'] = quarantine_nulls[
    'anomaly_columns'
].apply(
    lambda x: 'B13'
    if any(col in x for col in date_cols)
    else 'B08'
)
```

```python
# Quarantine all the rows with nulls CSV
quarantine_nulls.to_csv(
    'quarantine_nullsPT.csv',
    index=False
)
```

```python
#Checking invalid amounts

amount_cols = [
    'amount',
    'commission_amount',
    'acquisition_expense_amount',
    'administration_expense_amount'
          # add/remove columns as needed
]

for col in amount_cols:
    invalid_amounts = premium_transactions.loc[
        pd.to_numeric(premium_transactions[col], errors='coerce').isna()
        & premium_transactions[col].notna(),
        col
    ]

    print(f"\nInvalid values in {col}:")
    print(invalid_amounts.value_counts())
```

```python
# Convert monetary columns to numeric

money_cols = [
    'amount',
    'commission_amount',
    'acquisition_expense_amount',
    'administration_expense_amount'
]

for col in money_cols:
    premium_transactions[col] = pd.to_numeric(
        premium_transactions[col],
        errors='coerce'
    )
```

```python
# Clean monetary columns

money_cols = [
    'amount',
    'commission_amount',
    'acquisition_expense_amount',
    'administration_expense_amount'
]

for col in money_cols:
    premium_transactions[col] = (
        premium_transactions[col]
        .astype(str)
        .str.replace('(', '', regex=False)
        .str.replace(')', '', regex=False)
        .replace(['--', 'not recorded'], pd.NA)
        .str.strip()
    )
```

```python
# Remove brackets from monetary columns

money_cols = [
    'amount',
    'commission_amount',
    'acquisition_expense_amount',
    'administration_expense_amount'
]

for col in money_cols:
    policy_endorsements[col] = (
        policy_endorsements[col]
        .astype(str)
        .str.replace('(', '', regex=False)
        .str.replace(')', '', regex=False)
    )
 
```

```python
# Removing , and . on the sum_insured

money_cols = [
    'amount',
    'commission_amount',
    'acquisition_expense_amount',
    'administration_expense_amount'
]

for col in money_cols:

    # Convert European format: 1.234,56 -> 1234.56
    mask = premium_transactions[col].astype(str).str.contains(
        r'\.\d{3},\d{2}$',
        na=False
    )

    premium_transactions.loc[mask, col] = (
        premium_transactions.loc[mask, col]
        .str.replace('.', '', regex=False)
        .str.replace(',', '.', regex=False)
    )

    # Convert US format: 1,234.56 -> 1234.56
    premium_transactions[col] = (
        premium_transactions[col]
        .str.replace(',', '', regex=False)
    )
```

```python
# Remove currency symbols/codes (ZAR, zar, R) from monetary columns
money_cols = [
    'amount',
    'commission_amount',
    'acquisition_expense_amount',
    'administration_expense_amount'
]

for col in money_cols:
    premium_transactions[col] = (
        premium_transactions[col]
        .astype(str)
        .str.replace('ZAR', '', regex=False)
        .str.replace('R', '', regex=False)
        .str.replace('zar', '', regex=False)
        .str.strip()
    )
```

```python
# Checking invalid dates
date_cols = [
    'transaction_date',
    'cover_start_date',
    'cover_end_date'
]

for col in date_cols:
    tmp_date = pd.to_datetime(
        premium_transactions[col],
        format='mixed',
        errors='coerce'
    )

    print(f"\nInvalid {col} values:")
    print(
        premium_transactions.loc[
            tmp_date.isna() & premium_transactions[col].notna(),
            col
        ].value_counts(dropna=False)
    )
```

```python
#Quarantining the invalid dates 
date_cols = [
    'transaction_date',
    'cover_start_date',
    'cover_end_date'
]

quarantine_list = []

for col in date_cols:
    parsed = pd.to_datetime(
        premium_transactions[col],
        format='mixed',
        errors='coerce'
    )

    invalid_mask = (
        parsed.isna() &
        premium_transactions[col].notna()
    )

    tmp = premium_transactions[invalid_mask].copy()
    tmp['quarantine_reason'] = f'Invalid {col}'
    quarantine_list.append(tmp)

quarantine_invalid_dates = pd.concat(quarantine_list).drop_duplicates()

quarantine_invalid_dates.to_csv(
    'Quarantine/quarantine_invalid_datesPT.csv',
    index=False
)

print(f"Quarantined {len(quarantine_invalid_dates)} rows")
```

```python
# Converting date to datetime data type
date_cols = [
    'transaction_date',
    'cover_start_date',
    'cover_end_date'
]

for col in date_cols:
    premium_transactions[col] = pd.to_datetime(
        premium_transactions[col],
        format='mixed',
        errors='coerce'
    )
```
