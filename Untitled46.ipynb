# Cleaning Customers #

```python
import pandas as pd

customers = pd.read_csv("Raw Data/santam_raw_customers.csv")

customers.head()
```

```python
customers.info()
```

```python
# Change date_of_birth_or_registration from OBJ to DATE format:
import pandas as pd

customers = pd.read_csv("Raw Data/santam_raw_customers.csv")

# data["date_of_birth_or_registration"] = pd.to_datetime(
#     data["date_of_birth_or_registration"],
#     errors="coerce"
# )

print(customers["date_of_birth_or_registration"].dtype)
```

```python
# Check data in 'province' column:
print(customers["province"].unique())
```

```python
# Replace all Nan values in 'province' to 'Unspecified'
customers["province"] = customers["province"].fillna("Unspecified")
```

```python
# Check data in 'customer_type':
print(customers["customer_type"].unique())
```

```python
# Capitalise data in 'customer_type' column:
customers["customer_type"] = customers["customer_type"].str.capitalize()
```

```python
# Check data in 'source_system':
print(customers["source_system"].unique())
```

```python
customers.head()
```

```python
# Correct wording format in 'source_system' column:
customers['source_system'] = customers['source_system'].replace(
    'LegacyShortTermB',
    'Legacy Short Term B')

customers['source_system'] = customers['source_system'].replace(
    'LegacyShortTermA',
    'Legacy Short Term A')

customers['source_system'] = customers['source_system'].replace(
    'BrokerPortal',
    'Broker Portal')

customers['source_system'] = customers['source_system'].replace(
    'AcquiredBook',
    'Acquired Book')

```

```python
customers.head()
```

```python
# Cleam date foramt in 'date_of_birth_or_registration' column:
# customers['date_of_birth_or_registration'] = customers['date_of_birth_or_registration'].replace(
#     '/',
#     '-')
```

```python
# Replace NaT with 'Unspecified' in 'date_of_birth_or_registration' column:
customers['date_of_birth_or_registration'] = customers['date_of_birth_or_registration'].fillna('Unspecified')
```

```python
customers.tail(10)
```

```python
cols = customers.columns
for col in cols:
    if "date" in col:
      print(customers[col].describe())
```

```python
# Create a copy for the cleaned dataset
santam_clean_customers = customers.copy()

# Save cleaned dataset
santam_clean_customers.to_csv(
    "Clean Data/santam_clean_customers.csv",
    index=False
)
```
