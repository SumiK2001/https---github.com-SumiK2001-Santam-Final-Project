# Policies

```python
import pandas as pd

policies = pd.read_csv("Raw Data/santam_raw_policies.csv")

```

```python
policies.head()
```

```python
policies.info()
```

```python
#putting a placeholder, Since these fields are not applicable for non-motor policies

cols = [
    'vehicle_year',
    'vehicle_registration',
    'vehicle_vin_placeholder',
    'postal_area',
    'province'
]

policies[cols] = policies[cols].replace('', pd.NA)
```

```python
# Removing , and . on the annual_premium

mask = policies['annual_premium'].str.contains(
    r'\.\d{3},\d{2}', na=False
)
 
policies.loc[mask, 'annual_premium'] = (
    policies.loc[mask, 'annual_premium']
    .str.replace('.', '', regex=False)
    .str.replace(',', '.', regex=False)
)
 
policies['annual_premium'] = (
    policies['annual_premium']
    .str.replace(',', '', regex=False)
)
```

```python
# Removing , and . on the sum_insured

mask = policies['sum_insured'].str.contains(
    r'\.\d{3},\d{2}', na=False
)
 
policies.loc[mask, 'sum_insured'] = (
    policies.loc[mask, 'sum_insured']
    .str.replace('.', '', regex=False)
    .str.replace(',', '.', regex=False)
)
 
policies['sum_insured'] = (
    policies['sum_insured']
    .str.replace(',', '', regex=False)
)
```

```python
# Checking invalid amounts 

invalid_amounts = policies.loc[
    pd.to_numeric(policies['annual_premium'], errors='coerce').isna()
    & policies['annual_premium'].notna(),
    'annual_premium'
]

invalid_amounts.value_counts()
```

```python
#Replacing the invalid amount wit NaN
policies['annual_premium'] = policies['annual_premium'].replace(
    ['--', 'not recorded'],
    pd.NA
)

policies['sum_insured'] = policies['sum_insured'].replace(
    ['--', 'not recorded'],
    pd.NA
)
```

```python
# CONVERTING ANNUAL_PREMIUM AND SUM_INSURED TO FLOAT 
policies['annual_premium'] = pd.to_numeric(
    policies['annual_premium'],
    errors='coerce'
)

policies['sum_insured'] = pd.to_numeric(
    policies['sum_insured'],
    errors='coerce'
)
```

```python
#Removing CURRENCY in the money
policies['annual_premium'] = (
    policies['annual_premium']
    .astype(str)
    .str.replace('ZAR', '', regex=False)
    .str.replace('R', '', regex=False)
    .str.strip()
)

policies['sum_insured'] = (
    policies['sum_insured']
    .astype(str)
    .str.replace('ZAR', '', regex=False)
    .str.replace('R', '', regex=False)
    .str.strip()
)
```

```python
# CSV for all the invalid amounts '--', 'not recorded'
quarantine_amounts = policies[
    policies['annual_premium'].isin(['--', 'not recorded'])
    |
    policies['sum_insured'].isin(['--', 'not recorded'])
]

quarantine_amounts.to_csv(
    "Quarantine/quarantine_invalid_amounts.csv",
    index=False
)
```

```python
#Checking invalid amounts

invalid_amounts = policies.loc[
    pd.to_numeric(policies['annual_premium'], errors='coerce').isna()
    & policies['annual_premium'].notna(),
    'annual_premium'
]

invalid_amounts.value_counts()
```

```python
# Converting Vehicle  Year to int
policies['vehicle_year'] = policies['vehicle_year'].astype('Int64')
```

```python
# Replacing INFORCE wit In Force in Policy status
policies['policy_status'] = policies['policy_status'].replace({
    'INFORCE': 'IN FORCE'
})
policies.policy_status.unique()
```

```python
# Fixing inconsistancies on Product Line and Policy Status
# Remove spaces and standardize capitalization
cols = ['product_line', 'policy_status']

for col in cols:
    policies[col] = (
        policies[col]
        .str.strip()
        .str.title()
    )

```

```python
# Coveerting Date to date type
policies['policy_start_date'] = pd.to_datetime(
    policies['policy_start_date'],
    format='mixed'
)

policies['policy_end_date'] = pd.to_datetime(
    policies['policy_end_date'],
    format='mixed'
)
policies.info()
```

```python
#Checking invalid dates 

tmp_start = pd.to_datetime(
    policies['policy_start_date'],
    format='mixed',
    errors='coerce'
)

tmp_end = pd.to_datetime(
    policies['policy_end_date'],
    format='mixed',
    errors='coerce'
)

print("Invalid policy_start_date values:")
print(
    policies.loc[
        tmp_start.isna(),
        'policy_start_date'
    ].value_counts(dropna=False)
)

print("\nInvalid policy_end_date values:")
print(
    policies.loc[
        tmp_end.isna(),
        'policy_end_date'
    ].value_counts(dropna=False)
)
 # WHAT i GOT AS RESULTS

# Invalid policy_start_date values:
# policy_start_date
# NaN           77
# 31/31/2024    75
# not-a-date    67
# 2024-19-42    60
# Name: count, dtype: int64

# Invalid policy_end_date values:
# policy_end_date
# 31/31/2024    81
# not-a-date    77
# 2024-19-42    71
# NaN           67
# Name: count, dtype: int64
```

```python
#THE QUARANTINE CSV FOR ALL THE INVALID DATES
# Invalid dates in either column
mask = tmp_start.isna() | tmp_end.isna()

# Create quarantine dataset
quarantine_dates = policies.loc[mask].copy()

# Save quarantine file
quarantine_dates.to_csv(
    "Quarantine/quarantine_all_invalid_dates.csv",
    index=False
)
```

```python
# Rplace invalid dates 

# Replace invalid start dates with NaT
policies.loc[
    tmp_start.isna(),
    'policy_start_date'
] = pd.NaT

# Replace invalid end dates with NaT
policies.loc[
    tmp_end.isna(),
    'policy_end_date'
] = pd.NaT
```

```python
# vehicle year that is over that the year we in 

policies[policies['vehicle_year'] > 2026]['vehicle_year'].value_counts().sort_index()
```

```python
# Standardise product_line values by removing extra spaces and converting to Title Case
policy_endorsements['product_line'] = (
    policy_endorsements['product_line']
    .str.strip()
    .str.title()
)

# Check the unique values after standardisation
policy_endorsements['product_line'].unique()
```

```python
#Number of unique values per column
distinct_values = policy_endorsements.nunique()

print(distinct_values)
```

```python
# CONVERTING SUM_INSURED TO FLOAT 

policy_endorsements['sum_insured'] = pd.to_numeric(
    policy_endorsements['sum_insured'],
    errors='coerce'
)
```

```python
# Removing , and . on the sum_insured

mask = policy_endorsements['sum_insured'].str.contains(
    r'\.\d{3},\d{2}', na=False
)
 
policy_endorsements.loc[mask, 'sum_insured'] = (
    policy_endorsements.loc[mask, 'sum_insured']
    .str.replace('.', '', regex=False)
    .str.replace(',', '.', regex=False)
)
 
policy_endorsements['sum_insured'] = (
    policy_endorsements['sum_insured']
    .str.replace(',', '', regex=False)
)
```

```python
#Replacing the invalid amount wit NaN

policy_endorsements['sum_insured'] = policy_endorsements['sum_insured'].replace(
    ['--', 'not recorded'],
    pd.NA
)
```

```python
# CSV for all the invalid amounts '--', 'not recorded'
quarantine_amounts = policy_endorsements[
    policy_endorsements['sum_insured'].isin(['--', 'not recorded'])
]

quarantine_amounts.to_csv(
    "Quarantine/quarantine_invalid_amountsPE.csv",
    index=False
)
```

```python
#Removing CURRENCY in the money

policy_endorsements['sum_insured'] = (
    policy_endorsements['sum_insured']
    .astype(str)
    .str.replace('ZAR', '', regex=False)
    .str.replace('R', '', regex=False)
    .str.replace('zar', '', regex=False)
    .str.strip()
)
```

```python
#Checking invalid amounts

invalid_amounts = policy_endorsements.loc[
    pd.to_numeric(policy_endorsements['sum_insured'], errors='coerce').isna()
    & policy_endorsements['sum_insured'].notna(),
    'sum_insured'
]

invalid_amounts.value_counts()
```

```python
#Checking invalid dates 

tmp_effective = pd.to_datetime(
    policy_endorsements['effective_date'],
    format='mixed',
    errors='coerce'
)

print("Invalid effective_date values:")
print(
    policy_endorsements.loc[
        tmp_effective.isna(),
        'effective_date'
    ].value_counts(dropna=False)
)
```

```python
#THE QUARANTINE CSV FOR ALL THE INVALID DATES
# Invalid dates in either column
mask = tmp_effective.isna()

# Create quarantine dataset
quarantine_dates = policy_endorsements.loc[mask].copy()

# Save quarantine file
quarantine_dates.to_csv(
    "Quarantine/quarantine_invalid_datesPE.csv",
    index=False
)
```

```python
# Replace invalid dates 

# Replace invalid effective_date with NaT
policy_endorsements.loc[
    tmp_start.isna(),
    'effective_date'
] = pd.NaT
```

```python
# Converting Vehicle  Year to int
policy_endorsements['vehicle_year'] = policy_endorsements['vehicle_year'].astype('Int64')
```

```python
# Coveerting Date to date type
policy_endorsements['effective_date'] = pd.to_datetime(
    policy_endorsements['effective_date'],
    format='mixed'
)
```
